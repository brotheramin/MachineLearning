```python

# ================================================================
# BRISBANE RIVER - INTELLIGENT WATER QUALITY EARLY WARNING SYSTEM
# ================================================================
#
# Dataset: Brisbane River - Colmslie monitoring station
# Prediction horizon: 30 minutes
#
# MODELS:
# 1. Persistence Baseline
# 2. Random Forest Regression
# 3. HistGradientBoosting Regression
# 4. Isolation Forest Anomaly Detection
# 5. Gaussian Naive Bayes Risk Classification
# 6. Bayesian Expected-Cost Decision Support
#
# VALIDATION:
# - Chronological 80/20 split
# - Training-only imputation
# - Training-only risk thresholds
# - No future measurements used as model inputs
#
# NOTE:
# Risk categories are constructed research labels.
# They are NOT official regulatory water-quality classifications.
# Anomalies do not identify specific pollutants or their sources.
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.ensemble import (
    RandomForestRegressor,
    HistGradientBoostingRegressor,
    IsolationForest
)

from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)


# ================================================================
# 2. UPLOAD DATASET
# ================================================================

print("=" * 70)
print("BRISBANE RIVER WATER QUALITY EARLY WARNING SYSTEM")
print("=" * 70)

print("\nUpload your CSV dataset.")

uploaded = files.upload()

if not uploaded:
    raise ValueError("No dataset was uploaded.")

file_name = next(iter(uploaded.keys()))

df = pd.read_csv(file_name)

print("\nDataset loaded successfully.")
print("Original shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())


# ================================================================
# 3. DATA CLEANING
# ================================================================

df.columns = df.columns.str.strip()

if "Timestamp" not in df.columns:
    raise ValueError("Timestamp column was not found.")

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    errors="coerce"
)

df = df.dropna(subset=["Timestamp"])

# Remove quality columns without assuming their code meanings.

columns_to_remove = [
    col for col in df.columns
    if "quality" in col.lower()
]

if "Record number" in df.columns:
    columns_to_remove.append("Record number")

df = df.drop(
    columns=columns_to_remove,
    errors="ignore"
)

column_mapping = {
    "Average Water Speed": "Speed",
    "Average Water Direction": "Direction",
    "Temperature": "Temp",
    "Dissolved Oxygen": "DO",
    "Dissolved Oxygen (%Saturation)": "DO_Saturation",
    "Specific Conductance": "Conductance"
}

df = df.rename(columns=column_mapping)

measurement_variables = [
    "Speed",
    "Direction",
    "Temp",
    "DO",
    "Chlorophyll",
    "DO_Saturation",
    "pH",
    "Salinity",
    "Conductance",
    "Turbidity"
]

measurement_variables = [
    col for col in measurement_variables
    if col in df.columns
]

for col in measurement_variables:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

measurement_variables = [
    col for col in measurement_variables
    if df[col].notna().any()
]

if len(measurement_variables) < 3:
    raise ValueError(
        "Not enough valid measurement variables."
    )

df = df[
    ["Timestamp"] + measurement_variables
]

df = df.sort_values("Timestamp")

df = (
    df.groupby("Timestamp", as_index=False)
    .mean(numeric_only=True)
)

print("\nDataset after cleaning:", df.shape)


# ================================================================
# 4. CIRCULAR ENCODING OF WATER DIRECTION
# ================================================================

if "Direction" in df.columns:

    radians = np.deg2rad(df["Direction"])

    df["Direction_sin"] = np.sin(radians)

    df["Direction_cos"] = np.cos(radians)

    df = df.drop(columns=["Direction"])


# ================================================================
# 5. RESAMPLE TO 30-MINUTE INTERVALS
# ================================================================

df = df.set_index("Timestamp")

df = df.resample("30min").mean()

df = df.reset_index()

print("\nDataset after 30-minute resampling:", df.shape)

print(
    "\nDate range:",
    df["Timestamp"].min(),
    "to",
    df["Timestamp"].max()
)

print("\nMissing data percentages:")

print(
    (df.isna().mean() * 100).round(2)
)


# ================================================================
# 6. IDENTIFY FORECASTING TARGETS
# ================================================================

target_variables = [
    "Temp",
    "DO",
    "Chlorophyll",
    "Speed",
    "DO_Saturation",
    "pH",
    "Salinity",
    "Conductance",
    "Turbidity"
]

target_variables = [
    col for col in target_variables
    if col in df.columns and df[col].notna().any()
]

if not target_variables:
    raise ValueError("No forecasting targets are available.")

print("\nForecasting targets:")
print(target_variables)


# ================================================================
# 7. CHRONOLOGICAL TRAIN/TEST SPLIT
# ================================================================

split_position = int(len(df) * 0.80)

split_time = df.iloc[split_position]["Timestamp"]

next_timestamp = df["Timestamp"].shift(-1)

valid_next = (
    next_timestamp - df["Timestamp"]
) == pd.Timedelta(minutes=30)

train_mask = (
    (df["Timestamp"] < split_time)
    &
    (next_timestamp < split_time)
    &
    valid_next
)

test_mask = (
    (df["Timestamp"] >= split_time)
    &
    valid_next
)

print("\nTraining cutoff:", split_time)

print("Training observations:", train_mask.sum())

print("Testing observations:", test_mask.sum())

if train_mask.sum() == 0 or test_mask.sum() == 0:
    raise ValueError("Training or testing data is empty.")


# ================================================================
# 8. TEMPORAL FEATURE ENGINEERING
# ================================================================

feature_variables = [
    col for col in [
        "Temp",
        "DO",
        "Chlorophyll",
        "Speed",
        "DO_Saturation",
        "pH",
        "Salinity",
        "Conductance",
        "Turbidity",
        "Direction_sin",
        "Direction_cos"
    ]
    if col in df.columns and df[col].notna().any()
]

X = pd.DataFrame(index=df.index)

# Historical measurement lags.

for col in feature_variables:

    for lag in [0, 1, 2, 3, 6, 12]:

        X[f"{col}_lag_{lag}"] = (
            df[col].shift(lag)
        )

    # Rolling averages and standard deviations.

    for window in [3, 6, 12]:

        X[f"{col}_rolling_mean_{window}"] = (
            df[col].rolling(
                window=window,
                min_periods=2
            ).mean()
        )

        X[f"{col}_rolling_std_{window}"] = (
            df[col].rolling(
                window=window,
                min_periods=2
            ).std()
        )

# Time features.

hour = df["Timestamp"].dt.hour

day_of_year = df["Timestamp"].dt.dayofyear

day_of_week = df["Timestamp"].dt.dayofweek

X["Hour_sin"] = np.sin(
    2 * np.pi * hour / 24
)

X["Hour_cos"] = np.cos(
    2 * np.pi * hour / 24
)

X["DayOfYear_sin"] = np.sin(
    2 * np.pi * day_of_year / 365.25
)

X["DayOfYear_cos"] = np.cos(
    2 * np.pi * day_of_year / 365.25
)

X["DayOfWeek_sin"] = np.sin(
    2 * np.pi * day_of_week / 7
)

X["DayOfWeek_cos"] = np.cos(
    2 * np.pi * day_of_week / 7
)

print("\nNumber of engineered features:", X.shape[1])


# ================================================================
# 9. DEFINE FORECASTING MODELS
# ================================================================

forecast_models = {

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=15,
        min_samples_split=5,
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    ),

    "HistGradientBoosting": HistGradientBoostingRegressor(
        max_iter=200,
        learning_rate=0.05,
        max_leaf_nodes=15,
        l2_regularization=1.0,
        random_state=42
    )
}


# ================================================================
# 10. FORECASTING AND EVALUATION
# ================================================================

forecast_results = []

forecast_metrics = []

trained_models = {}

def evaluate_regression(y_true, y_pred):

    return {
        "MAE": mean_absolute_error(
            y_true,
            y_pred
        ),

        "RMSE": np.sqrt(
            mean_squared_error(
                y_true,
                y_pred
            )
        ),

        "R2": r2_score(
            y_true,
            y_pred
        )
    }


for target in target_variables:

    print("\n" + "=" * 65)
    print("FORECASTING:", target)
    print("=" * 65)

    # Next 30-minute measurement.

    y = df[target].shift(-1)

    # IMPORTANT FIX:
    # Training requires a valid future target.
    # Testing requires both a valid future target AND
    # a valid current measurement for the persistence baseline.

    target_train_mask = (
        train_mask
        & y.notna()
    )

    target_test_mask = (
        test_mask
        & y.notna()
        & df[target].notna()
    )

    X_train = X.loc[
        target_train_mask
    ].copy()

    X_test = X.loc[
        target_test_mask
    ].copy()

    y_train = y.loc[
        target_train_mask
    ].copy()

    y_test = y.loc[
        target_test_mask
    ].copy()

    if len(X_train) < 100 or len(X_test) < 20:

        print("Insufficient valid data. Skipping.")
        continue

    # Keep features with at least one training observation.

    usable_features = X_train.columns[
        X_train.notna().any()
    ]

    X_train = X_train[usable_features].copy()

    X_test = X_test[usable_features].copy()

    # Training-only median imputation.

    medians = X_train.median()

    X_train = X_train.fillna(medians)

    X_test = X_test.fillna(medians)

    # Remove any columns still containing missing values.

    usable_features = X_train.columns[
        X_train.notna().all()
    ]

    X_train = X_train[usable_features]

    X_test = X_test[usable_features]

    if X_train.shape[1] == 0:

        print("No usable input features. Skipping.")
        continue

    y_test_values = y_test.values

    # ------------------------------------------------------------
    # PERSISTENCE BASELINE
    # ------------------------------------------------------------

    persistence_predictions = df.loc[
        target_test_mask,
        target
    ].values

    # Both current and future values are valid because of
    # target_test_mask. Additional protection against NaNs.

    valid_baseline = (
        np.isfinite(y_test_values)
        &
        np.isfinite(persistence_predictions)
    )

    if valid_baseline.sum() == 0:

        print("No valid baseline observations. Skipping.")
        continue

    y_test_values = y_test_values[valid_baseline]

    persistence_predictions = (
        persistence_predictions[valid_baseline]
    )

    # Apply the same valid rows to model inputs.

    X_test = X_test.iloc[
        np.where(valid_baseline)[0]
    ].copy()

    baseline_metrics = evaluate_regression(
        y_test_values,
        persistence_predictions
    )

    forecast_metrics.append({

        "Target": target,

        "Model": "Persistence Baseline",

        **baseline_metrics

    })

    print("\nPersistence baseline:")
    print(baseline_metrics)

    target_output = pd.DataFrame({

        "Timestamp": df.loc[
            target_test_mask,
            "Timestamp"
        ].values[valid_baseline],

        "Target": target,

        "Actual": y_test_values,

        "Persistence": persistence_predictions

    })

    # ------------------------------------------------------------
    # TRAIN MACHINE LEARNING MODELS
    # ------------------------------------------------------------

    for model_name, model in forecast_models.items():

        print("\nTraining:", model_name)

        model.fit(
            X_train,
            y_train
        )

        predictions = model.predict(
            X_test
        )

        metrics = evaluate_regression(
            y_test_values,
            predictions
        )

        forecast_metrics.append({

            "Target": target,

            "Model": model_name,

            **metrics

        })

        target_output[model_name] = predictions

        trained_models[
            (target, model_name)
        ] = model

        print("MAE:", round(metrics["MAE"], 5))

        print("RMSE:", round(metrics["RMSE"], 5))

        print("R2:", round(metrics["R2"], 5))

    forecast_results.append(target_output)


if not forecast_results:

    raise ValueError(
        "No forecasting models produced results."
    )

forecast_results = pd.concat(
    forecast_results,
    ignore_index=True
)

forecast_metrics = pd.DataFrame(
    forecast_metrics
)


# ================================================================
# 11. FORECASTING PERFORMANCE SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("FORECASTING PERFORMANCE")
print("=" * 70)

print(
    forecast_metrics.sort_values(
        ["Target", "RMSE"]
    ).to_string(index=False)
)

best_forecast_models = (
    forecast_metrics
    .sort_values(["Target", "RMSE"])
    .groupby("Target")
    .first()
    .reset_index()
)

print("\nBEST MODEL FOR EACH TARGET:")

print(
    best_forecast_models[
        ["Target", "Model", "MAE", "RMSE", "R2"]
    ].to_string(index=False)
)


# ================================================================
# 12. FORECASTING VISUALIZATIONS
# ================================================================

for target in forecast_results["Target"].unique():

    subset = forecast_results[
        forecast_results["Target"] == target
    ].head(300)

    plt.figure(figsize=(14, 5))

    plt.plot(
        subset["Timestamp"],
        subset["Actual"],
        label="Actual",
        linewidth=2
    )

    plt.plot(
        subset["Timestamp"],
        subset["Persistence"],
        label="Persistence",
        alpha=0.7
    )

    for model_name in forecast_models.keys():

        if model_name in subset.columns:

            plt.plot(
                subset["Timestamp"],
                subset[model_name],
                label=model_name,
                alpha=0.8
            )

    plt.title(
        f"Actual vs Predicted - {target}"
    )

    plt.xlabel("Timestamp")
    plt.ylabel(target)

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


# ================================================================
# 13. ANOMALY DETECTION - ISOLATION FOREST
# ================================================================

print("\n" + "=" * 70)
print("ANOMALY DETECTION")
print("=" * 70)

anomaly_train = X.loc[
    train_mask
].copy()

anomaly_test = X.loc[
    test_mask
].copy()

anomaly_features = anomaly_train.columns[
    anomaly_train.notna().any()
]

anomaly_train = anomaly_train[
    anomaly_features
].copy()

anomaly_test = anomaly_test[
    anomaly_features
].copy()

anomaly_medians = anomaly_train.median()

anomaly_train = anomaly_train.fillna(
    anomaly_medians
)

anomaly_test = anomaly_test.fillna(
    anomaly_medians
)

usable_anomaly_features = anomaly_train.columns[
    anomaly_train.notna().all()
]

anomaly_train = anomaly_train[
    usable_anomaly_features
]

anomaly_test = anomaly_test[
    usable_anomaly_features
]

if anomaly_train.shape[1] == 0:

    raise ValueError(
        "No usable anomaly detection features."
    )

isolation_model = IsolationForest(
    n_estimators=200,
    contamination=0.03,
    random_state=42,
    n_jobs=-1
)

isolation_model.fit(
    anomaly_train
)

anomaly_predictions = isolation_model.predict(
    anomaly_test
)

anomaly_scores = isolation_model.decision_function(
    anomaly_test
)

anomaly_results = pd.DataFrame({

    "Timestamp": df.loc[
        test_mask,
        "Timestamp"
    ].values,

    "AnomalyScore": anomaly_scores,

    "AnomalyFlag": np.where(
        anomaly_predictions == -1,
        "Anomaly",
        "Normal"
    )

})

print("\nAnomaly distribution:")

print(
    anomaly_results["AnomalyFlag"].value_counts()
)

print(
    "\nAnomaly percentage:",
    round(
        (anomaly_predictions == -1).mean() * 100,
        2
    ),
    "%"
)

plt.figure(figsize=(14, 5))

plt.plot(
    anomaly_results["Timestamp"],
    anomaly_results["AnomalyScore"]
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title("Isolation Forest Anomaly Scores")

plt.xlabel("Timestamp")
plt.ylabel("Anomaly Score")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ================================================================
# 14. CONSTRUCT FUTURE WATER-QUALITY RISK LABELS
# ================================================================

print("\n" + "=" * 70)
print("CONSTRUCTING FUTURE RISK LABELS")
print("=" * 70)

baseline_data = df.loc[
    train_mask
].copy()

risk_rules = {}

if (
    "DO" in df.columns
    and baseline_data["DO"].notna().any()
):

    risk_rules["DO"] = {

        "type": "low",

        "threshold": baseline_data["DO"].quantile(0.25)

    }

if (
    "Turbidity" in df.columns
    and baseline_data["Turbidity"].notna().any()
):

    risk_rules["Turbidity"] = {

        "type": "high",

        "threshold": baseline_data["Turbidity"].quantile(0.75)

    }

if (
    "Chlorophyll" in df.columns
    and baseline_data["Chlorophyll"].notna().any()
):

    risk_rules["Chlorophyll"] = {

        "type": "high",

        "threshold": baseline_data["Chlorophyll"].quantile(0.75)

    }

if (
    "pH" in df.columns
    and baseline_data["pH"].notna().any()
):

    risk_rules["pH"] = {

        "type": "range",

        "lower": baseline_data["pH"].quantile(0.10),

        "upper": baseline_data["pH"].quantile(0.90)

    }

if len(risk_rules) < 3:

    raise ValueError(
        "At least three risk indicators are required."
    )

print("\nTraining-derived risk thresholds:")

for variable, rule in risk_rules.items():

    print(variable, rule)

future_measurements = df[
    list(risk_rules.keys())
].shift(-1)

risk_score = pd.Series(
    0,
    index=df.index,
    dtype="int64"
)

available_score = pd.Series(
    0,
    index=df.index,
    dtype="int64"
)

for variable, rule in risk_rules.items():

    future_value = future_measurements[
        variable
    ]

    if rule["type"] == "low":

        condition = (
            future_value < rule["threshold"]
        )

    elif rule["type"] == "high":

        condition = (
            future_value > rule["threshold"]
        )

    else:

        condition = (
            (future_value < rule["lower"])
            |
            (future_value > rule["upper"])
        )

    available = (
        future_value.notna()
        &
        valid_next
    )

    risk_score += (
        condition.fillna(False)
        & available
    ).astype(int)

    available_score += (
        available.astype(int)
    )

minimum_available = min(
    3,
    len(risk_rules)
)

valid_risk = (
    valid_next
    &
    (available_score >= minimum_available)
)

df["FutureRisk"] = pd.Series(
    index=df.index,
    dtype="object"
)

df.loc[
    valid_risk & (risk_score == 0),
    "FutureRisk"
] = "Low"

df.loc[
    valid_risk & (risk_score == 1),
    "FutureRisk"
] = "Moderate"

df.loc[
    valid_risk & (risk_score >= 2),
    "FutureRisk"
] = "High"

risk_states = [
    "Low",
    "Moderate",
    "High"
]

print("\nFuture risk distribution:")

print(
    df["FutureRisk"].value_counts(
        dropna=True
    )
)


# ================================================================
# 15. PREPARE BAYESIAN CLASSIFICATION DATA
# ================================================================

print("\n" + "=" * 70)
print("BAYESIAN PROBABILISTIC DECISION SUPPORT")
print("=" * 70)

risk_feature_names = []

for variable in [
    "DO",
    "Turbidity",
    "Chlorophyll",
    "pH",
    "Temp",
    "Salinity"
]:

    for lag in [0, 1]:

        feature_name = f"{variable}_lag_{lag}"

        if feature_name in X.columns:

            risk_feature_names.append(
                feature_name
            )

for col in [
    "Hour_sin",
    "Hour_cos",
    "DayOfYear_sin",
    "DayOfYear_cos"
]:

    if col in X.columns:

        risk_feature_names.append(col)

if len(risk_feature_names) < 2:

    raise ValueError(
        "Not enough Bayesian classification features."
    )

risk_train_mask = (
    train_mask
    & df["FutureRisk"].notna()
)

risk_test_mask = (
    test_mask
    & df["FutureRisk"].notna()
)

X_risk_train = X.loc[
    risk_train_mask,
    risk_feature_names
].copy()

X_risk_test = X.loc[
    risk_test_mask,
    risk_feature_names
].copy()

y_risk_train = df.loc[
    risk_train_mask,
    "FutureRisk"
].copy()

y_risk_test = df.loc[
    risk_test_mask,
    "FutureRisk"
].copy()

risk_medians = X_risk_train.median()

X_risk_train = X_risk_train.fillna(
    risk_medians
)

X_risk_test = X_risk_test.fillna(
    risk_medians
)

usable_risk_features = X_risk_train.columns[
    X_risk_train.notna().all()
]

X_risk_train = X_risk_train[
    usable_risk_features
]

X_risk_test = X_risk_test[
    usable_risk_features
]

if X_risk_train.shape[1] == 0:

    raise ValueError(
        "No usable Bayesian classifier features."
    )

if len(y_risk_train) == 0 or len(y_risk_test) == 0:

    raise ValueError(
        "Bayesian training or testing dataset is empty."
    )

print("\nTraining risk distribution:")

print(y_risk_train.value_counts())

print("\nTesting risk distribution:")

print(y_risk_test.value_counts())


# ================================================================
# 16. TRAIN GAUSSIAN NAIVE BAYES
# ================================================================

bayesian_model = GaussianNB(
    var_smoothing=1e-8
)

bayesian_model.fit(
    X_risk_train,
    y_risk_train
)

risk_predictions = bayesian_model.predict(
    X_risk_test
)

posterior_array = bayesian_model.predict_proba(
    X_risk_test
)

model_risk_states = list(
    bayesian_model.classes_
)

posterior = pd.DataFrame(
    posterior_array,
    columns=model_risk_states,
    index=X_risk_test.index
)

for state in risk_states:

    if state not in posterior.columns:

        posterior[state] = 0.0

posterior = posterior[risk_states]

print("\nBayesian classifier trained successfully.")


# ================================================================
# 17. EVALUATE BAYESIAN CLASSIFICATION
# ================================================================

risk_accuracy = accuracy_score(
    y_risk_test,
    risk_predictions
)

risk_balanced_accuracy = balanced_accuracy_score(
    y_risk_test,
    risk_predictions
)

risk_macro_f1 = f1_score(
    y_risk_test,
    risk_predictions,
    labels=risk_states,
    average="macro",
    zero_division=0
)

print("\nBAYESIAN CLASSIFICATION PERFORMANCE")

print(
    "Accuracy:",
    round(risk_accuracy, 4)
)

print(
    "Balanced accuracy:",
    round(risk_balanced_accuracy, 4)
)

print(
    "Macro F1-score:",
    round(risk_macro_f1, 4)
)

print("\nClassification report:")

print(
    classification_report(
        y_risk_test,
        risk_predictions,
        labels=risk_states,
        zero_division=0
    )
)

risk_confusion = confusion_matrix(
    y_risk_test,
    risk_predictions,
    labels=risk_states
)

print("\nConfusion matrix:")

print(
    pd.DataFrame(
        risk_confusion,
        index=risk_states,
        columns=risk_states
    )
)


# ================================================================
# 18. CONFUSION MATRIX VISUALIZATION
# ================================================================

plt.figure(figsize=(7, 5))

plt.imshow(
    risk_confusion,
    interpolation="nearest"
)

plt.title("Bayesian Classifier Confusion Matrix")

plt.xlabel("Predicted Risk")

plt.ylabel("Actual Risk")

plt.xticks(
    range(3),
    risk_states
)

plt.yticks(
    range(3),
    risk_states
)

for i in range(3):

    for j in range(3):

        plt.text(
            j,
            i,
            risk_confusion[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()

plt.tight_layout()

plt.show()


# ================================================================
# 19. BAYESIAN EXPECTED-COST DECISION SUPPORT
# ================================================================

actions = [
    "Routine Monitoring",
    "Increase Monitoring",
    "Confirmatory Sampling",
    "Investigate Potential Source"
]

# Illustrative cost assumptions only.
# These are not official operational costs.

cost_matrix = pd.DataFrame(

    [
        [0, 2, 5, 10],
        [5, 2, 4, 7],
        [20, 10, 4, 8]
    ],

    index=risk_states,

    columns=actions

)

print("\nIllustrative decision cost matrix:")

print(cost_matrix)

decision_results = []

for idx, probabilities in posterior.iterrows():

    expected_costs = {}

    for action in actions:

        expected_cost = sum(

            probabilities[state]
            * cost_matrix.loc[state, action]

            for state in risk_states

        )

        expected_costs[action] = expected_cost

    recommended_action = min(
        expected_costs,
        key=expected_costs.get
    )

    decision_results.append({

        "RecommendedAction": recommended_action,

        "ExpectedCost": expected_costs[
            recommended_action
        ]

    })


# ================================================================
# 20. CREATE DECISION SUPPORT OUTPUT
# ================================================================

decision_results = pd.DataFrame(
    decision_results,
    index=posterior.index
)

decision_output = pd.DataFrame({

    "Timestamp": df.loc[
        risk_test_mask,
        "Timestamp"
    ].values,

    "ActualRisk": y_risk_test.values,

    "PredictedRisk": risk_predictions,

    "LowProbability": posterior["Low"].values,

    "ModerateProbability": posterior["Moderate"].values,

    "HighProbability": posterior["High"].values,

    "RecommendedAction": decision_results[
        "RecommendedAction"
    ].values,

    "ExpectedCost": decision_results[
        "ExpectedCost"
    ].values

})

print("\nDecision-support results:")

print(decision_output.head(15))


# ================================================================
# 21. COMBINE ANOMALY AND DECISION SUPPORT RESULTS
# ================================================================

decision_output = decision_output.merge(
    anomaly_results,
    on="Timestamp",
    how="left"
)

print("\nCombined results:")

print(decision_output.head(15))


# ================================================================
# 22. RECOMMENDED ACTION VISUALIZATION
# ================================================================

action_counts = decision_output[
    "RecommendedAction"
].value_counts()

plt.figure(figsize=(10, 5))

action_counts.plot(
    kind="bar"
)

plt.title("Recommended Monitoring Actions")

plt.xlabel("Monitoring Action")

plt.ylabel("Number of Observations")

plt.xticks(
    rotation=25,
    ha="right"
)

plt.tight_layout()

plt.show()


# ================================================================
# 23. POSTERIOR RISK PROBABILITY VISUALIZATION
# ================================================================

plt.figure(figsize=(14, 6))

plt.plot(
    decision_output["Timestamp"],
    decision_output["LowProbability"],
    label="Low Risk"
)

plt.plot(
    decision_output["Timestamp"],
    decision_output["ModerateProbability"],
    label="Moderate Risk"
)

plt.plot(
    decision_output["Timestamp"],
    decision_output["HighProbability"],
    label="High Risk"
)

plt.title("Bayesian Posterior Water-Quality Risk Probabilities")

plt.xlabel("Timestamp")

plt.ylabel("Probability")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ================================================================
# 24. RANDOM FOREST FEATURE IMPORTANCE
# ================================================================

print("\n" + "=" * 70)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 70)

for target in target_variables:

    key = (
        target,
        "Random Forest"
    )

    if key not in trained_models:

        continue

    model = trained_models[key]

    importance = pd.DataFrame({

        "Feature": model.feature_names_in_,

        "Importance": model.feature_importances_

    }).sort_values(
        "Importance",
        ascending=False
    )

    print("\nTarget:", target)

    print(
        importance.head(10).to_string(index=False)
    )

    top_features = importance.head(10)

    plt.figure(figsize=(9, 5))

    plt.barh(
        top_features["Feature"][::-1],
        top_features["Importance"][::-1]
    )

    plt.title(
        f"Top 10 Feature Importances - {target}"
    )

    plt.xlabel("Importance")

    plt.tight_layout()

    plt.show()


# ================================================================
# 25. EXPORT RESULTS
# ================================================================

forecast_results_file = (
    "Colmslie_Forecasting_Results.csv"
)

forecast_metrics_file = (
    "Colmslie_Forecasting_Metrics.csv"
)

anomaly_file = (
    "Colmslie_Anomaly_Detection.csv"
)

decision_file = (
    "Colmslie_Bayesian_Decision_Support.csv"
)

forecast_results.to_csv(
    forecast_results_file,
    index=False
)

forecast_metrics.to_csv(
    forecast_metrics_file,
    index=False
)

anomaly_results.to_csv(
    anomaly_file,
    index=False
)

decision_output.to_csv(
    decision_file,
    index=False
)

print("\nAll CSV files exported successfully.")


# ================================================================
# 26. DOWNLOAD RESULTS
# ================================================================

print("\nDownloading results...")

for output_file in [
    forecast_results_file,
    forecast_metrics_file,
    anomaly_file,
    decision_file
]:

    files.download(output_file)


# ================================================================
# 27. FINAL PROJECT SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("PROJECT COMPLETED")
print("=" * 70)

print("\nDataset: Brisbane River - Colmslie")

print("Prediction horizon: 30 minutes")

print("Chronological split: 80% training / 20% testing")

print("\nForecasting models:")
print("- Persistence Baseline")
print("- Random Forest")
print("- HistGradientBoosting")

print("\nAnomaly detection:")
print("- Isolation Forest")

print("\nBayesian decision support:")
print("- Gaussian Naive Bayes")
print("- Posterior risk probabilities")
print("- Expected-cost monitoring decisions")

print("\nFORECASTING PERFORMANCE:")

print(
    forecast_metrics.sort_values(
        ["Target", "RMSE"]
    ).to_string(index=False)
)

print("\nBAYESIAN CLASSIFICATION PERFORMANCE:")

print("Accuracy:", round(risk_accuracy, 4))

print(
    "Balanced accuracy:",
    round(risk_balanced_accuracy, 4)
)

print("Macro F1:", round(risk_macro_f1, 4))

print("\nANOMALY DISTRIBUTION:")

print(
    anomaly_results["AnomalyFlag"].value_counts()
)

print("\nRECOMMENDED ACTIONS:")

print(
    decision_output["RecommendedAction"].value_counts()
)

print("\nGenerated files:")
print("- Colmslie_Forecasting_Results.csv")
print("- Colmslie_Forecasting_Metrics.csv")
print("- Colmslie_Anomaly_Detection.csv")
print("- Colmslie_Bayesian_Decision_Support.csv")

print("\nBRISBANE RIVER PROJECT FINISHED.")
```
